## Hasil

Artefak model dan metrik ada di `artifacts/`. Dataset ada di `data/processed/`. Salin ke Google Drive dengan sel berikut.


```python
import shutil, pathlib
src=pathlib.Path('/content/AIFX-Sentinel')
dst=pathlib.Path('/content/drive/MyDrive/AIFX_XAUUSD')
dst.mkdir(parents=True,exist_ok=True)
for name in ['data/processed','artifacts']:
    s=src/name
    if s.exists():
        d=dst/name
        if d.exists(): shutil.rmtree(d)
        shutil.copytree(s,d)
print('Saved to',dst)
```
